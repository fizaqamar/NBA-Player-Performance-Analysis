#**8) 🛠️ Feature Engineering 🧠**
Feature engineering is the process of creating new meaningful features from existing data to help machine learning models better understand patterns in the dataset.

In this project, simple basketball efficiency-based features are created using existing numerical columns.

**🎯 Why Feature Engineering is Useful**
Raw statistics alone may not always capture team efficiency. By combining related statistics, we can derive new features that represent performance quality, not just quantity.

These engineered features may improve:

*    model accuracy

*    interpretability

*    learning efficiency


**🔢 Assist-to-Turnover Ratio (AST / TO) 🧮**
The Assist-to-Turnover Ratio measures how well a team converts possessions into assists while minimizing mistakes.

A higher value generally indicates:

* better ball movement

* fewer turnovers

* more efficient offense

**🧪 Create AST/TO Ratio**


```python
nba_df["AST_TO_RATIO"] = nba_df["AST"] / nba_df["TO"]
nba_df["AST_TO_RATIO"].head()
```

**⚠️ Handling Division by Zero**

Some games may have zero turnovers, which can cause infinite values. These values must be handled before further analysis.

**🧪 Replace infinite values with NaN**

```python
nba_df["AST_TO_RATIO"].replace([np.inf, -np.inf], np.nan, inplace=True)
```

**🧼 Handle missing values in engineered feature**

```python
nba_df["AST_TO_RATIO"].fillna(nba_df["AST_TO_RATIO"].median(), inplace=True)
```

**🔥 Scoring Efficiency Feature (FGM / FGA)**
Another useful feature is Scoring Efficiency, which shows how effectively a team converts shot attempts into made shots.

**🧪 Create scoring efficiency feature**

```python
nba_df["SCORING_EFFICIENCY"] = nba_df["FGM"] / nba_df["FGA"]
nba_df["SCORING_EFFICIENCY"].head()
```

**🧼 Handle invalid values**


```python
nba_df["SCORING_EFFICIENCY"].replace([np.inf, -np.inf], np.nan, inplace=True)
nba_df["SCORING_EFFICIENCY"].fillna(nba_df["SCORING_EFFICIENCY"].median(), inplace=True)
```

**✅ Verify New Features**

```python
nba_df[["AST_TO_RATIO", "SCORING_EFFICIENCY"]].describe()
```

# **12) 🤖 Machine Learning Preparation 🧠**

Before training machine learning models, the dataset must be properly prepared.
This includes selecting features, defining the target variable, splitting the data into training and testing sets, and scaling numerical features.

**🎯 Defining the Target Variable (Game Outcome)**



The goal of this project is to predict game outcome (Win/Loss).
In this dataset, the target variable is:

RESULT

* 1 → Win

* 0 → Loss

```python
target = "RESULT"
```

 **🧪 Selecting Feature Columns**

Only numerical columns are used as input features for machine learning models.
The target variable is excluded from the feature set.

```python
X = (
    nba_df
    .select_dtypes(include=["int64", "float64"])
    .drop(columns=[target], errors="ignore")
)

y = nba_df[target]
```

The parameter `errors="ignore"` is used to prevent errors in case the target
column is not present in the selected numeric feature set. This makes the
feature-selection step more robust and avoids unexpected crashes.

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

###**📏 Feature Scaling (Standardization)**

KNN uses distances. If one column has big values (PTS) and another has small values (FG_PCT), the big one dominates. Scaling makes features comparable.

Feature scaling is applied to ensure that all numerical features are on the same
scale. This is especially important for distance-based algorithms such as KNN.

StandardScaler transforms the data so that each feature has:
- Mean ≈ 0
- Standard Deviation ≈ 1

# **13) **🧠** K-Nearest Neighbors (KNN) Model Training ⚙️**
K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for
classification. It predicts the class of a data point based on the majority
class of its nearest neighbors.

Since KNN is a distance-based algorithm, feature scaling is essential.
In this section, the KNN model is trained to predict the game outcome
(Win/Loss) using the scaled NBA statistics.

```python
# 1) Replace inf with NaN (some features may create inf due to division)
X_clean = X.replace([np.inf, -np.inf], np.nan)

# 2) Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y, test_size=0.2, random_state=42, stratify=y
)

# 3) Pipeline: Imputer -> Scaler -> KNN
knn_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier(n_neighbors=7))
])

# 4) Train
knn_pipe.fit(X_train, y_train)
```

Before training the KNN model, it is useful to inspect whether any numerical
features contain missing values. This step helps identify columns that may
cause errors in distance-based algorithms such as KNN.

```python
X.isna().sum().sort_values(ascending=False).head(20)
```

```python
#Predict game outcomes on test data
y_pred_knn = knn_pipe.predict(X_test)
```

```python
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
```

📌 This output shows how well the KNN model performed on unseen data.


The accuracy score represents the percentage of correct predictions made
by the KNN model on the test dataset.
A higher accuracy indicates better model performance.

#**14) Confusion Matrix 📊**
The confusion matrix shows how the KNN model performed:

- True Positives (TP): Correctly predicted wins
- True Negatives (TN): Correctly predicted losses
- False Positives (FP): Losses predicted as wins
- False Negatives (FN): Wins predicted as losses

This helps understand not only accuracy, but also the types of mistakes
the model makes.
\**bold text**

```python
# Create confusion matrix
cm = confusion_matrix(y_test, y_pred_knn)

# Display confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Loss", "Win"]
)

disp.plot()
plt.title("Confusion Matrix - KNN Model")
plt.show()
```

# **📈 Logistic Regression Model**
Logistic Regression is a supervised machine learning algorithm commonly used for
binary classification problems.

In this project, Logistic Regression is used as a **baseline model** to compare
its performance with the KNN model for predicting game outcomes (Win/Loss).

Although the name contains the word "regression", Logistic Regression is used
for classification tasks.

**🧠 Why Use a Pipeline?**



A pipeline ensures that preprocessing steps such as:
- handling missing values
- scaling features

are applied automatically before training the model.

This makes the model training process cleaner, safer, and less error-prone.

```python
#Create Logistic Regression Pipeline
logreg_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000))
])
```

```python
#train the Logistic Regression Model
logreg_pipe.fit(X_train, y_train)
```

```python
#Make Predictions
y_pred_logreg = logreg_pipe.predict(X_test)
```

```python
#Accuracy Output 📊
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_logreg))
```

The accuracy score represents the proportion of correctly predicted game
outcomes. This value is used to compare the performance of Logistic Regression
with other models such as KNN.

```python
#Confusion Matrix 📋
cm_logreg = confusion_matrix(y_test, y_pred_logreg)

ConfusionMatrixDisplay(
    confusion_matrix=cm_logreg,
    display_labels=["Loss", "Win"]
).plot()

plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

# **🌳 Decision Tree Classifier**
A Decision Tree is a supervised machine learning algorithm that makes decisions
by splitting data into branches based on feature values.

It is easy to understand and interpret, as it mimics human decision-making.
Unlike KNN and Logistic Regression, Decision Trees do **not require feature scaling**.


🧠 Why Use a Decision Tree?

Decision Trees:
- Handle non-linear relationships
- Do not require feature scaling
- Are easy to visualize and explain

They are useful for understanding which features influence predictions.

```python
#Create Decision Tree Pipeline
tree_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("model", DecisionTreeClassifier(
        random_state=42,
        max_depth=6
    ))
])
```

📌 max_depth=6 prevents overfitting and keeps the tree simple.

```python
#Train the Decision Tree Model
tree_pipe.fit(X_train, y_train)
```

```python
#Make Predictions
y_pred_tree = tree_pipe.predict(X_test)
```

```python
#Accuracy Output 📊

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_tree))
```

**🧠 Accuracy Interpretation**

The accuracy score shows how well the Decision Tree model predicts game outcomes
on unseen test data. This result can be compared with KNN and Logistic Regression
to evaluate relative performance.

```python
#Confusion Matrix 📋
cm_tree = confusion_matrix(y_test, y_pred_tree)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Loss", "Win"]
).plot()

plt.title("Confusion Matrix - Decision Tree")
plt.show()
```

 📋 Confusion Matrix Explanation

The confusion matrix shows the classification performance of the Decision Tree:

- Correct predictions appear on the diagonal
- Incorrect predictions appear off the diagonal

This visualization helps identify whether the model is biased toward predicting
wins or losses.
